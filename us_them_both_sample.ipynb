```python
import pandas as pd
import re

# ========== 1. LOAD DATASET ==========
file_path = "D:/melikeh/full_748k_tweets.csv" 
print("Loading dataset...")
df = pd.read_csv(file_path)
print(f"✅ Dataset loaded with {len(df):,} rows.")

# ========== 2. DEFINE SEPARATE PATTERNS FOR 'WE' AND 'THEY' ==========
# Pattern for 'we': ما, ماها, ما ها
pattern_we = r'\bما(ها|ها)?\b'
# Pattern for 'they': آنها, اونا, اونها, آن ها, اونا, اون ها
pattern_they = r'\b(آنها|اونا|اونها|آن ها|اونا|اون ها)\b'

# ========== 3. FIND TWEETS THAT HAVE BOTH 'WE' AND 'THEY' ==========
print("Searching for tweets that have both 'we' and 'they'...")

has_we = df['text'].str.contains(pattern_we, na=False, regex=True)
has_they = df['text'].str.contains(pattern_they, na=False, regex=True)

df_both = df[has_we & has_they].copy()
print(f"✅ {len(df_both):,} tweets found that have both 'we' and 'they'.")

# ========== 4. EXTRACT EXACT WORDS FOUND FOR EACH TWEET (OPTIONAL) ==========
def extract_we_words(text):
    return re.findall(pattern_we, text)

def extract_they_words(text):
    return re.findall(pattern_they, text)

df_both['we_words'] = df_both['text'].apply(extract_we_words)
df_both['they_words'] = df_both['text'].apply(extract_they_words)

# ========== 5. SHOW SAMPLE TWEETS ==========
print("\n📝 Sample tweets containing both 'we' and 'they':")
for idx, row in df_both.head(10).iterrows():
    text = row['text']
    we_found = ', '.join(set(row['we_words']))
    they_found = ', '.join(set(row['they_words']))
    print(f"   Tweet: {text[:150]}...")
    print(f"   'We' found: {we_found}")
    print(f"   'They' found: {they_found}")
    print("-" * 70)

# ========== 6. SAVE RESULTS ==========
output_file = "us_them_both_sample.csv"
df_both[['text', 'we_words', 'they_words']].to_csv(output_file, index=False, encoding='utf-8-sig')
print(f"\n🎉 File '{output_file}' saved with {len(df_both)} tweets.")
```
